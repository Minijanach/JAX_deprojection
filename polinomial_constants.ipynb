# Polynomial Fit
This notebook allows you to input some points and fits a 3rd-degree polynomial to them.

```python
import numpy as np
import matplotlib.pyplot as plt

# Input points
points = np.array([
    [0, 0.1],
    [1, 0.3],
    [2, 0.6],
    [3, 0.8],
    [4, 0.5],
    [5, 0.2],
    [6, 0.3],
    [7, 0.6],
    [8, 0.7],
    [9, 0.9]
])
x = points[:, 0]
y = points[:, 1]

# Fit a 3rd-degree polynomial
A = np.vstack([x**3, x**2, x, np.ones(len(x))]).T
coefs = np.linalg.lstsq(A, y, rcond=None)[0]
print("Coefficients:", coefs)

# Plot the points and the polynomial fit
x_fit = np.linspace(min(x), max(x), 100)
y_fit = coefs[0]*x_fit**3 + coefs[1]*x_fit**2 + coefs[2]*x_fit + coefs[3]

plt.scatter(x, y, color='red', label='Data Points')
plt.plot(x_fit, y_fit, label='3rd Degree Polynomial Fit')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.title('3rd Degree Polynomial Fit')
plt.show()
```
