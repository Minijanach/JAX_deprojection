```python
import numpy as np
import matplotlib.pyplot as plt

# Define the cubic polynomial
def p(x):
    return 0.0905 * x**3 - 0.0274 * x**2 + 0.0492 * x + 0.5276

# Generate values for x
x_values = np.linspace(0, 10, 400)  # Adjust the range of x-values as needed

# Calculate corresponding y-values for the cubic polynomial
y_values = p(x_values)

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x_values, y_values, label=r'$p(x) = 0.0905x^3 - 0.0274x^2 + 0.0492x + 0.5276$', color='blue')
plt.scatter([0.1, 1, 3.16, 10], [0.55, 0.6, 0.625, 0.8], color='red', zorder=5, label="Data Points")

# Annotate the points
plt.text(0.1, 0.55, "(0.1, 0.55)", fontsize=12, verticalalignment='bottom', horizontalalignment='right')
plt.text(1, 0.6, "(1, 0.6)", fontsize=12, verticalalignment='bottom', horizontalalignment='right')
plt.text(3.16, 0.625, "(3.16, 0.625)", fontsize=12, verticalalignment='bottom', horizontalalignment='right')
plt.text(10, 0.8, "(10, 0.8)", fontsize=12, verticalalignment='bottom', horizontalalignment='right')

# Add labels and title
plt.xlabel('x')
plt.ylabel('p(x)')
plt.title('Plot of the Cubic Polynomial')

# Display the legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the data points
x_data = np.array([0.1, 1., 3.16, 10])
y_data = np.array([0.55, 0.6, 0.625, 0.8])

# Fit a cubic polynomial to the data points (degree 3)
coefficients = np.polyfit(x_data, y_data, 3)

# Print the coefficients
print("Cubic Polynomial Coefficients:", coefficients)

# Define the polynomial function using the coefficients
p = np.poly1d(coefficients)

# Generate x values for plotting (smooth curve)
x_values = np.linspace(0, 10, 400)
y_values = p(x_values)

# Plot the data points and the fitted cubic polynomial
plt.figure(figsize=(8, 6))
plt.plot(x_values, y_values, label="Fitted Cubic Polynomial", color='blue')
plt.scatter(x_data, y_data, color='red', zorder=5, label="Data Points")

# Annotate the points
for i, (x, y) in enumerate(zip(x_data, y_data)):
    plt.text(x, y, f"({x}, {y})", fontsize=12, verticalalignment='bottom', horizontalalignment='right')

# Add labels and title
plt.xlabel('x')
plt.ylabel('p(x)')
plt.title('Cubic Polynomial Fit to Data Points')
#plt.xscale('log')

# Display the legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()

```
