```python
def func(x,y,z):
    return str(x)+str(y)+str(z)
```

```python
x = np.array([1,2,3])
y = np.array([1,2,3])
z = np.array([1,2,3])
```

```python
zer = np.zeros((3,3,3))

for i in range(3):
    for j in range(3):
        for k in range(3):
            zer[i,j,k] = func(x[i],y[j],z[k])
```

```python
zer[0,:,:]
```

```python
zer
```

```python
zer_sum = np.sum(zer, axis=2)
```

```python
zer_sum
```

```python
correct_zer = zer_sum.T

plt.imshow(correct_zer, origin='lower')
plt.colorbar()
```
